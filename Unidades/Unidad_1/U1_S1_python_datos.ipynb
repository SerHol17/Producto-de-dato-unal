## 1. Estructuras de datos en Python

Antes de manipular datos con librerías especializadas, necesitamos dominar las estructuras nativas de Python. Cada una tiene un propósito distinto y elegir la correcta impacta directamente en la eficiencia y claridad del código.

### 1.1 Listas

Una lista es una colección **ordenada** y **mutable** de elementos. Es la estructura más versátil de Python.

```python
# ============================================================
# LISTAS — colección ordenada y mutable
# ============================================================

# Crear una lista con diferentes tipos de datos
# Python permite mezclar tipos, pero en la práctica es mejor mantener un solo tipo
ventas_mensuales = [1500, 2300, 1800, 2100, 2500, 1900]

# Acceder al primer elemento (índice 0)
print("Primera venta:", ventas_mensuales[0])  # 1500

# Acceder al último elemento (índice -1)
print("Última venta:", ventas_mensuales[-1])  # 1900

# Slicing — obtener un subconjunto de la lista
# [inicio:fin] → incluye inicio, excluye fin
primer_trimestre = ventas_mensuales[0:3]
print("Primer trimestre:", primer_trimestre)  # [1500, 2300, 1800]
```

```python
# ============================================================
# OPERACIONES COMUNES CON LISTAS
# ============================================================

# Agregar un elemento al final
ventas_mensuales.append(2800)
print("Después de append:", ventas_mensuales)

# Insertar en una posición específica
# insert(posición, valor)
ventas_mensuales.insert(0, 1200)  # Agrega 1200 al inicio
print("Después de insert:", ventas_mensuales)

# Eliminar un elemento por valor
ventas_mensuales.remove(1200)  # Elimina la primera ocurrencia de 1200
print("Después de remove:", ventas_mensuales)

# Longitud de la lista
print("Total de meses registrados:", len(ventas_mensuales))

# Funciones de agregación
print("Venta máxima:", max(ventas_mensuales))
print("Venta mínima:", min(ventas_mensuales))
print("Total de ventas:", sum(ventas_mensuales))
print("Promedio:", sum(ventas_mensuales) / len(ventas_mensuales))
```

```python
# ============================================================
# LIST COMPREHENSIONS — forma compacta de crear listas
# ============================================================

# Sintaxis: [expresión for variable in iterable if condición]

# Ejemplo 1: convertir ventas de miles a unidades
ventas_en_unidades = [venta * 1000 for venta in ventas_mensuales]
print("En unidades:", ventas_en_unidades)

# Ejemplo 2: filtrar solo ventas por encima del promedio
promedio = sum(ventas_mensuales) / len(ventas_mensuales)
ventas_altas = [venta for venta in ventas_mensuales if venta > promedio]
print("Ventas por encima del promedio:", ventas_altas)

# Ejemplo 3: crear etiquetas para cada mes
etiquetas = [f"Mes {i+1}: ${venta}" for i, venta in enumerate(ventas_mensuales)]
# enumerate() nos da el índice (i) y el valor (venta) en cada iteración
for etiqueta in etiquetas:
    print(etiqueta)
```

### 1.2 Diccionarios

Un diccionario almacena pares **clave-valor**. Es la estructura ideal cuando necesitamos asociar datos con un identificador único — como un registro de base de datos.

```python
# ============================================================
# DICCIONARIOS — pares clave:valor
# ============================================================

# Crear un diccionario que represente un producto
# Cada clave es un string, los valores pueden ser de cualquier tipo
producto = {
    "nombre": "Dashboard de Ventas",   # string
    "version": 2.1,                     # float
    "activo": True,                     # booleano
    "usuarios": ["ana", "carlos"],      # lista dentro del diccionario
    "metricas": {                       # diccionario anidado
        "visitas_diarias": 150,
        "tiempo_carga_seg": 2.3
    }
}

# Acceder a un valor por su clave
print("Nombre:", producto["nombre"])

# Acceder a un valor anidado
print("Visitas diarias:", producto["metricas"]["visitas_diarias"])

# Acceso seguro con .get() — retorna None si la clave no existe
# Evita el error KeyError
print("Descripción:", producto.get("descripcion", "No definida"))
```

```python
# ============================================================
# OPERACIONES CON DICCIONARIOS
# ============================================================

# Agregar o modificar un par clave-valor
producto["descripcion"] = "Dashboard interactivo para el equipo comercial"
producto["version"] = 2.2  # Modificar un valor existente

# Eliminar una clave
del producto["activo"]

# Recorrer claves y valores
# .items() retorna tuplas (clave, valor)
print("=== Contenido del producto ===")
for clave, valor in producto.items():
    print(f"  {clave}: {valor}")

# Verificar si una clave existe
print("\n¿Tiene versión?", "version" in producto)  # True
print("¿Tiene precio?", "precio" in producto)      # False
```

```python
# ============================================================
# LISTA DE DICCIONARIOS — estructura muy común en datos
# ============================================================

# Esto es esencialmente lo que un DataFrame representa
# Cada diccionario es una fila, cada clave es una columna
empleados = [
    {"nombre": "Ana García",    "area": "Datos",     "salario": 4500},
    {"nombre": "Carlos López",  "area": "Backend",   "salario": 4800},
    {"nombre": "María Ruiz",    "area": "Datos",     "salario": 5200},
    {"nombre": "Pedro Martín",  "area": "Frontend",  "salario": 4200},
]

# Filtrar empleados del área de Datos
equipo_datos = [emp for emp in empleados if emp["area"] == "Datos"]
print("Equipo de datos:")
for emp in equipo_datos:
    print(f"  {emp['nombre']} — ${emp['salario']}")

# Calcular salario promedio
salarios = [emp["salario"] for emp in empleados]
print(f"\nSalario promedio: ${sum(salarios) / len(salarios):.0f}")
```

### 1.3 Tuplas

Una tupla es como una lista pero **inmutable** — una vez creada, no se puede modificar. Se usa para datos que no deben cambiar, como coordenadas, configuraciones fijas o claves compuestas.

```python
# ============================================================
# TUPLAS — colección ordenada e inmutable
# ============================================================

# Coordenadas de Manizales — no deberían cambiar
coordenadas_manizales = (5.0689, -75.5174)

# Desempaquetar una tupla en variables individuales
latitud, longitud = coordenadas_manizales
print(f"Latitud: {latitud}, Longitud: {longitud}")

# Las tuplas son inmutables — esto generaría un error:
# coordenadas_manizales[0] = 5.07  # TypeError!

# Uso práctico: retornar múltiples valores desde una función
def calcular_estadisticas(datos):
    """Retorna una tupla con (mínimo, máximo, promedio)"""
    return min(datos), max(datos), sum(datos) / len(datos)

minimo, maximo, promedio = calcular_estadisticas(ventas_mensuales)
print(f"Mín: {minimo}, Máx: {maximo}, Promedio: {promedio:.1f}")
```

### 1.4 Sets

Un set es una colección **no ordenada** de elementos **únicos**. Perfecto para eliminar duplicados y hacer operaciones de conjuntos (unión, intersección, diferencia).

```python
# ============================================================
# SETS — colección de elementos únicos, no ordenada
# ============================================================

# Ejemplo: usuarios que accedieron a dos dashboards diferentes
usuarios_dashboard_ventas = {"ana", "carlos", "maría", "pedro", "ana"}  # "ana" aparece 2 veces
print("Usuarios únicos:", usuarios_dashboard_ventas)  # "ana" aparece solo 1 vez

usuarios_dashboard_ops = {"carlos", "luisa", "pedro", "jorge"}

# Unión — usuarios que accedieron a al menos un dashboard
todos = usuarios_dashboard_ventas | usuarios_dashboard_ops
print("Todos los usuarios:", todos)

# Intersección — usuarios que accedieron a ambos dashboards
en_ambos = usuarios_dashboard_ventas & usuarios_dashboard_ops
print("En ambos dashboards:", en_ambos)

# Diferencia — usuarios que solo usan el dashboard de ventas
solo_ventas = usuarios_dashboard_ventas - usuarios_dashboard_ops
print("Solo en ventas:", solo_ventas)

# Caso práctico: eliminar duplicados de una lista
ciudades_con_duplicados = ["Manizales", "Bogotá", "Medellín", "Bogotá", "Manizales"]
ciudades_unicas = list(set(ciudades_con_duplicados))
print("Ciudades únicas:", ciudades_unicas)
```

### Resumen: ¿Cuándo usar cada estructura?

| Estructura | Ordenada | Mutable | Duplicados | Caso de uso típico |
|---|---|---|---|---|
| **Lista** | ✅ | ✅ | ✅ | Colecciones generales, secuencias de datos |
| **Diccionario** | ✅* | ✅ | Claves únicas | Registros, configuraciones, mapeos |
| **Tupla** | ✅ | ❌ | ✅ | Coordenadas, retornos múltiples, claves compuestas |
| **Set** | ❌ | ✅ | ❌ | Eliminar duplicados, operaciones de conjuntos |

\* Los diccionarios mantienen orden de inserción desde Python 3.7+

```python
# ============================================================
# CREAR UN DATAFRAME DESDE UN DICCIONARIO
# ============================================================

# Cada clave del diccionario se convierte en una columna
# Cada valor es una lista con los datos de esa columna
# Todas las listas deben tener la misma longitud
datos_ventas = {
    "fecha":     ["2025-01-15", "2025-01-16", "2025-01-17", "2025-01-18", "2025-01-19"],
    "producto":  ["Dashboard",  "Reporte",    "Dashboard",  "API",        "Reporte"],
    "region":    ["Eje Cafetero", "Bogotá",   "Medellín",   "Bogotá",     "Eje Cafetero"],
    "unidades":  [15,            8,            22,           5,            12],
    "ingreso":   [4500.0,        2400.0,       6600.0,       3500.0,       3600.0]
}

# pd.DataFrame() convierte el diccionario en tabla
df = pd.DataFrame(datos_ventas)

# Mostrar el DataFrame completo
# En Jupyter, la última línea de una celda se renderiza automáticamente como tabla
df
```

### 2.3 Selección y filtrado

Seleccionar columnas y filtrar filas son las operaciones más frecuentes. Hay varias formas de hacerlo — veamos las más usadas.

```python
# ============================================================
# SELECCIONAR COLUMNAS
# ============================================================

# Una sola columna → retorna una Serie (columna individual)
print("Columna 'producto':")
print(df["producto"])
print(f"Tipo: {type(df['producto'])}")
```

```python
# Múltiples columnas → retorna un DataFrame
# Se pasa una lista de nombres de columnas
df[["producto", "ingreso"]]
```

### 2.4 Transformaciones

Transformar datos es el día a día: crear columnas nuevas, cambiar tipos, renombrar, ordenar.

```python
# ============================================================
# CREAR COLUMNAS NUEVAS
# ============================================================

# Columna calculada: precio unitario = ingreso / unidades
df["precio_unitario"] = df["ingreso"] / df["unidades"]

# Columna con operación sobre strings
# .str da acceso a métodos de string vectorizados
df["producto_upper"] = df["producto"].str.upper()

# Columna condicional con np.where
import numpy as np

# np.where(condición, valor_si_true, valor_si_false)
df["categoria"] = np.where(df["ingreso"] > 4000, "Alto", "Bajo")

df
```

```python
# ============================================================
# CONVERTIR TIPOS DE DATOS
# ============================================================

# La columna 'fecha' es un string — convertirla a datetime
# pd.to_datetime() reconoce automáticamente muchos formatos
df["fecha"] = pd.to_datetime(df["fecha"])

# Ahora podemos extraer componentes de la fecha
df["dia_semana"] = df["fecha"].dt.day_name()     # Nombre del día
df["mes"] = df["fecha"].dt.month                  # Número del mes

print("Tipos actualizados:")
print(df.dtypes)
print("\n")
df[["fecha", "dia_semana", "mes"]]
```

```python
# ============================================================
# RENOMBRAR COLUMNAS
# ============================================================

# .rename() con un diccionario {nombre_viejo: nombre_nuevo}
df_renombrado = df.rename(columns={
    "ingreso": "ingreso_total",
    "unidades": "cantidad"
})

# NOTA: por defecto, .rename() retorna una copia
# Para modificar el original: df.rename(..., inplace=True)
df_renombrado.columns.tolist()
```

```python
# ============================================================
# ORDENAR
# ============================================================

# Ordenar por una columna
# ascending=False → de mayor a menor
df.sort_values("ingreso", ascending=False)
```

```python
# Ordenar por múltiples columnas
# Primero por producto (A-Z), luego por ingreso (mayor a menor)
df.sort_values(["producto", "ingreso"], ascending=[True, False])
```

### 2.5 Agrupación y agregación

`.groupby()` es el equivalente a `GROUP BY` en SQL. Permite agrupar filas por una o más columnas y calcular estadísticas sobre cada grupo.

```python
# ============================================================
# GROUPBY — agrupar y agregar
# ============================================================

# Ingreso total por producto
# .groupby("columna") agrupa las filas
# ["ingreso"] selecciona la columna a agregar
# .sum() calcula la suma de cada grupo
ingreso_por_producto = df.groupby("producto")["ingreso"].sum()
print("Ingreso total por producto:")
print(ingreso_por_producto)
```

```python
# ============================================================
# MÚLTIPLES AGREGACIONES CON .agg()
# ============================================================

# .agg() permite aplicar varias funciones a la vez
# Se pasa un diccionario {columna: [funciones]}
resumen = df.groupby("producto").agg(
    ingreso_total=("ingreso", "sum"),        # suma del ingreso
    ingreso_promedio=("ingreso", "mean"),     # promedio del ingreso
    unidades_total=("unidades", "sum"),       # suma de unidades
    num_ventas=("ingreso", "count")           # conteo de registros
).reset_index()  # reset_index() convierte el índice de grupo en columna

resumen
```

```python
# ============================================================
# AGRUPAR POR MÚLTIPLES COLUMNAS
# ============================================================

# Ingreso por producto y región
df.groupby(["producto", "region"])["ingreso"].sum().reset_index()
```

### 2.6 Joins — combinar DataFrames

En la vida real, los datos viven en varias tablas. `merge()` nos permite combinarlas, igual que un `JOIN` en SQL.

```python
# ============================================================
# MERGE — combinar dos DataFrames
# ============================================================

# Tabla de productos con información adicional
info_productos = pd.DataFrame({
    "producto": ["Dashboard", "Reporte", "API"],
    "responsable": ["Ana", "Carlos", "María"],
    "costo_desarrollo": [10000, 5000, 15000]
})

print("Tabla de productos:")
print(info_productos)

# Merge — equivalente a LEFT JOIN en SQL
# on="producto" → columna en común
# how="left" → mantener todas las filas del DataFrame izquierdo
df_completo = df.merge(
    info_productos,   # DataFrame derecho
    on="producto",     # columna clave para el join
    how="left"         # tipo de join: left, right, inner, outer
)

print("\nDespués del merge:")
df_completo[["fecha", "producto", "ingreso", "responsable", "costo_desarrollo"]]
```

**Tipos de join disponibles:**

| `how=` | Comportamiento | Equivalente SQL |
|---|---|---|
| `"inner"` | Solo filas que coinciden en ambos DataFrames | `INNER JOIN` |
| `"left"` | Todas las filas del izquierdo + coincidencias del derecho | `LEFT JOIN` |
| `"right"` | Todas las filas del derecho + coincidencias del izquierdo | `RIGHT JOIN` |
| `"outer"` | Todas las filas de ambos DataFrames | `FULL OUTER JOIN` |

## 3. Funciones y modularidad

Escribir funciones nos permite reutilizar código, hacerlo legible y testeable. En un producto de datos, las funciones son los bloques con los que construimos pipelines.

```python
# ============================================================
# FUNCIONES BÁSICAS
# ============================================================

def calcular_margen(ingreso, costo):
    """
    Calcula el margen de ganancia como porcentaje.
    
    Parámetros:
        ingreso (float): ingreso total de la venta
        costo (float): costo asociado al producto
    
    Retorna:
        float: margen de ganancia en porcentaje
    """
    # Evitar división por cero
    if ingreso == 0:
        return 0.0
    
    margen = ((ingreso - costo) / ingreso) * 100
    return round(margen, 2)

# Probar la función
print(f"Margen: {calcular_margen(4500, 1000)}%")  # 77.78%
print(f"Margen: {calcular_margen(2400, 2400)}%")  # 0.0%
```

```python
# ============================================================
# PARÁMETROS POR DEFECTO Y *args/**kwargs
# ============================================================

def generar_reporte(df, titulo="Reporte", top_n=3, formato="simple"):
    """
    Genera un resumen de los top N productos por ingreso.
    
    Parámetros:
        df (DataFrame): datos de ventas
        titulo (str): título del reporte (por defecto: "Reporte")
        top_n (int): número de productos a mostrar (por defecto: 3)
        formato (str): 'simple' o 'detallado'
    """
    print(f"\n{'='*50}")
    print(f"  {titulo}")
    print(f"{'='*50}")
    
    # Agrupar por producto y sumar ingresos
    resumen = (df.groupby("producto")["ingreso"]
               .sum()
               .sort_values(ascending=False)
               .head(top_n))
    
    for producto, ingreso in resumen.items():
        if formato == "detallado":
            n_ventas = len(df[df["producto"] == producto])
            print(f"  {producto}: ${ingreso:,.0f} ({n_ventas} ventas)")
        else:
            print(f"  {producto}: ${ingreso:,.0f}")

# Llamada con valores por defecto
generar_reporte(df)

# Llamada personalizada
generar_reporte(df, titulo="Top Productos Enero", top_n=2, formato="detallado")
```

```python
# ============================================================
# FUNCIONES LAMBDA — funciones anónimas de una línea
# ============================================================

# Útiles para transformaciones rápidas dentro de pandas
# Sintaxis: lambda parámetros: expresión

# Ejemplo 1: aplicar una función a cada valor de una columna
# .apply() ejecuta la función fila por fila
df["ingreso_con_iva"] = df["ingreso"].apply(lambda x: x * 1.19)

# Ejemplo 2: categorizar con lógica más compleja
df["tamano_venta"] = df["unidades"].apply(
    lambda u: "Grande" if u > 15 else ("Mediana" if u > 8 else "Pequeña")
)

df[["producto", "unidades", "ingreso", "ingreso_con_iva", "tamano_venta"]]
```

```python
# ============================================================
# FUNCIONES QUE PROCESAN DATAFRAMES
# ============================================================

# Patrón muy útil en productos de datos:
# función que recibe un DataFrame, lo transforma y retorna otro

def preparar_datos(df_raw):
    """
    Pipeline de preparación de datos.
    Recibe datos crudos y retorna datos listos para análisis.
    """
    # Trabajar sobre una copia para no modificar el original
    df_prep = df_raw.copy()
    
    # 1. Asegurar que la fecha sea datetime
    if df_prep["fecha"].dtype == "object":
        df_prep["fecha"] = pd.to_datetime(df_prep["fecha"])
    
    # 2. Crear columna de precio unitario
    df_prep["precio_unit"] = df_prep["ingreso"] / df_prep["unidades"]
    
    # 3. Estandarizar nombres de producto a minúsculas
    df_prep["producto"] = df_prep["producto"].str.lower()
    
    # 4. Ordenar por fecha
    df_prep = df_prep.sort_values("fecha").reset_index(drop=True)
    
    return df_prep

# Usar la función
df_limpio = preparar_datos(df)
df_limpio
```

## Ejercicio integrador

Pongamos en práctica todo lo visto. El ejercicio simula un caso real: recibimos datos, los exploramos, limpiamos y generamos un resumen.

```python
# ============================================================
# EJERCICIO: Pipeline de análisis de ventas
# ============================================================

# 1. Crear un dataset más grande para simular datos reales
import random
random.seed(42)  # Semilla para reproducibilidad

n = 100  # 100 registros
productos = ["Dashboard", "Reporte", "API", "App Web"]
regiones = ["Eje Cafetero", "Bogotá", "Medellín", "Cali", "Barranquilla"]

datos_simulados = {
    "fecha": pd.date_range(start="2025-01-01", periods=n, freq="D"),
    "producto": [random.choice(productos) for _ in range(n)],
    "region": [random.choice(regiones) for _ in range(n)],
    "unidades": [random.randint(1, 50) for _ in range(n)],
    "ingreso": [round(random.uniform(500, 10000), 2) for _ in range(n)]
}

df_sim = pd.DataFrame(datos_simulados)
print(f"Dataset creado: {df_sim.shape[0]} filas, {df_sim.shape[1]} columnas")
df_sim.head()
```

```python
# 2. Exploración rápida
print("=== Info del dataset ===")
df_sim.info()
print("\n=== Estadísticas ===")
df_sim.describe()
```

```python
# 3. Transformaciones

# Precio unitario
df_sim["precio_unitario"] = round(df_sim["ingreso"] / df_sim["unidades"], 2)

# Mes como texto
df_sim["mes"] = df_sim["fecha"].dt.strftime("%B")  # Nombre completo del mes

# Categoría de venta
df_sim["categoria"] = pd.cut(
    df_sim["ingreso"],
    bins=[0, 3000, 7000, float("inf")],   # Rangos
    labels=["Bajo", "Medio", "Alto"]       # Etiquetas
)

df_sim.head()
```

```python
# 4. Análisis agrupado

# Resumen por producto
resumen_producto = df_sim.groupby("producto").agg(
    ventas=("ingreso", "count"),
    ingreso_total=("ingreso", "sum"),
    ingreso_promedio=("ingreso", "mean"),
    unidades_total=("unidades", "sum")
).round(2).sort_values("ingreso_total", ascending=False)

print("=== Resumen por Producto ===")
resumen_producto
```

```python
# 5. Guardar resultados

# Guardar el dataset procesado
df_sim.to_csv("ventas_procesadas.csv", index=False)

# Guardar el resumen
resumen_producto.to_csv("resumen_por_producto.csv")

print("✅ Archivos guardados:")
print("   - ventas_procesadas.csv")
print("   - resumen_por_producto.csv")
```
